```python
# ============================================================
# Q1: Build Your Personalized Knowledge Base
#
# Take your college roll number. Extract its digits.
# Build a pandas DataFrame with exactly 6 FAQ entries:
# 4 fixed entries and 2 entries constructed from your
# own roll number digits.
#
# Take the LAST TWO DIGITS of your roll number.
# For each digit d, compute:
#
# category = ["billing", "account", "general"][d % 3]
#
# Invent a realistic question-answer and 3 keywords
# for each personalized entry.
#
# Roll Number = 1024160087
# Last two digits = 8 and 7
#
# 8 % 3 = 2 -> general
# 7 % 3 = 1 -> account
#
# Output: Print your final 6-row DataFrame.
# ============================================================

import pandas as pd

# Roll number
roll_number = 1024160077

# Last two digits
last_two_digits = [7, 7]

# Category list given in the question
categories = ["billing", "account", "general"]

# Four fixed FAQ entries given in the question
fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fees",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

# Personalized entry for digit 8
digit = last_two_digits[0]
category = categories[digit % 3]

personalized_entry_1 = {
    "question": "how do i update my registered mobile number",
    "answer": "Go to Account Settings and update your registered mobile number.",
    "keywords": "update mobile number phone",
    "category": category
}

# Personalized entry for digit 7
digit = last_two_digits[1]
category = categories[digit % 3]

personalized_entry_2 = {
    "question": "how can i check my payment history",
    "answer": "Go to Billing and select Payment History to view your previous payments.",
    "keywords": "payment history transaction account",
    "category": category
}

# Combine all six entries
all_entries = fixed_entries + [
    personalized_entry_1,
    personalized_entry_2
]

# Create DataFrame
df = pd.DataFrame(all_entries)

# Print final DataFrame
print("Final 6-row DataFrame:")
print(df)
```

```python
# ============================================================
# Q2: Generate and Score a Hypothesis
#
# Implement a scoring function that takes a query string
# and returns all matching entries ranked by confidence.
#
# The score is based on the number of query words that
# match words present in the FAQ question or keywords.
# ============================================================

def score_query(query, df):

    # Convert query into lowercase words
    query_words = set(query.lower().split())

    results = []

    # Check every FAQ entry
    for index, row in df.iterrows():

        # Combine question and keywords
        text = (
            row["question"].lower()
            + " "
            + row["keywords"].lower()
        )

        text_words = set(text.split())

        # Find common words
        matched_words = query_words.intersection(text_words)

        # Confidence score
        score = len(matched_words)

        # Store only matching entries
        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score,
                "matched_words": ", ".join(
                    sorted(matched_words)
                )
            })

    # Rank entries from highest score to lowest score
    results.sort(
        key=lambda x: x["score"],
        reverse=True
    )

    return results


# Test the function
query = "how can i pay my fees"

results = score_query(query, df)

print("Query:", query)
print("\nMatching entries:")

for result in results:
    print(result)
```

```python
# ============================================================
# Q3: Write a function same_category(category_name, df)
# that returns all questions belonging to a given category.
#
# Call it using the category of one of the personalized
# entries from Q1, and print the result.
# ============================================================

def same_category(category_name, df):

    # Select rows belonging to the given category
    result = df[
        df["category"].str.lower() == category_name.lower()
    ]

    return result


# Use the category of the first personalized entry
category_name = personalized_entry_1["category"]

print("Category:", category_name)
print("\nQuestions belonging to this category:")

result = same_category(category_name, df)

print(result[["question", "category"]])
```

```python
# ============================================================
# Q6: Modify your Q2 scoring function so that if two or more
# entries tie for the highest score, it does not silently
# pick one.
#
# Instead, it prints ALL matching entries having the
# highest score.
#
# Demonstrate with one query that produces a tie
# (for example, a query matching both "fee" entries)
# and one that doesn't.
# ============================================================

def score_query_with_ties(query, df):

    # Convert query into lowercase words
    query_words = set(query.lower().split())

    results = []

    # Check every FAQ entry
    for index, row in df.iterrows():

        # Combine question and keywords
        text = (
            row["question"].lower()
            + " "
            + row["keywords"].lower()
        )

        text_words = set(text.split())

        # Find matching words
        matched_words = query_words.intersection(text_words)

        # Calculate score
        score = len(matched_words)

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score,
                "matched_words": ", ".join(
                    sorted(matched_words)
                )
            })

    # If there are no matches
    if len(results) == 0:
        return []

    # Find the highest score
    highest_score = max(
        result["score"]
        for result in results
    )

    # Return ALL entries having the highest score
    best_matches = [
        result
        for result in results
        if result["score"] == highest_score
    ]

    return best_matches


# ------------------------------------------------------------
# Demonstration 1: Query that produces a tie
# ------------------------------------------------------------

query = "fee"

best_matches = score_query_with_ties(query, df)

print("Query:", query)

if len(best_matches) == 0:

    print("No matching FAQ found.")

else:

    print("Highest score:", best_matches[0]["score"])
    print("Number of tied matches:", len(best_matches))
    print("\nAll best matching entries:")

    for result in best_matches:

        print("\nQuestion:", result["question"])
        print("Answer:", result["answer"])
        print("Category:", result["category"])
        print("Matched words:", result["matched_words"])
        print("Score:", result["score"])


# ------------------------------------------------------------
# Demonstration 2: Query that does NOT produce a tie
# ------------------------------------------------------------

query = "working hours"

best_matches = score_query_with_ties(query, df)

print("\n" + "=" * 50)
print("Second Query:", query)

if len(best_matches) == 0:

    print("No matching FAQ found.")

else:

    print("Highest score:", best_matches[0]["score"])
    print("\nBest matching entry:")

    for result in best_matches:

        print("\nQuestion:", result["question"])
        print("Answer:", result["answer"])
        print("Category:", result["category"])
        print("Matched words:", result["matched_words"])
        print("Score:", result["score"])
```
