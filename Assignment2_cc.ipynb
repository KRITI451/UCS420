## Ques 1 Take your roll number. Extract its individual digits, and multiply each digit by 10 to form your list L (10 numbers in list now).
### i. Print L
### ii. Add two numbers of your choice to L — one using append() and one using insert() at a specific position.
### Print L after each and explain in a comment what changed.
### iii. RemovetwoelementsfromL—oneusingremove()andoneusingpop().PrintLaftereach.
### iv. Sort L in ascending order using sort(), then descending order using sort(reverse=True). Print both.
### v. Use slicing to print the first three elements and the last three elements of L in a single line each.
### vi. Use list comprehension to create a new list containing only the elements of L that are greater than the average of L.

```python


roll_no = input("Enter your roll number: ")

# Extract digits and multiply each digit by 10
L = [int(digit) * 10 for digit in roll_no]

# i. Print L
print("Original List:", L)

# ii. Add two numbers
L.append(50)         
print("After append:", L)

L.insert(2, 70)      
print("After insert:", L)

# iii. Remove two elements
L.remove(50)         
print("After remove:", L)

L.pop(2)             
print("After pop:", L)

# iv. Sort ascending and descending
L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)

# v. Slicing
print("First three elements:", L[:3])
print("Last three elements:", L[-3:])

# vi. Elements greater than average
average = sum(L) / len(L)

greater_than_average = [x for x in L if x > average]

print("Average:", average)
print("Elements greater than average:", greater_than_average)
```

### 3. Set the random seed to your own roll number using random.seed(your_roll_number) before generating your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed produces a different list, so this cannot be copied from a classmate).
### i. Generate a list of 100 random numbers between 100 and 900 (inclusive).
### ii. Count and print all odd numbers in the list.
### iii. Countandprintallevennumbersinthelist.
### iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension.
### v. Print the number that occurs most frequently in your list, and how many times it occurs.

```python
import random
roll_no = int(input("Enter your roll number: "))
random.seed(roll_no)
numbers = [random.randint(100, 900) for _ in range(100)]
print("Random numbers:")
print(numbers)
odd_numbers = [x for x in numbers if x % 2 != 0]
print("Odd numbers:", odd_numbers)
print("Count of odd numbers:", len(odd_numbers))
even_numbers = [x for x in numbers if x % 2 == 0]
print("Even numbers:", even_numbers)
print("Count of even numbers:", len(even_numbers))
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


prime_numbers = [x for x in numbers if is_prime(x)]

print("Prime numbers:", prime_numbers)
print("Count of prime numbers:", len(prime_numbers))
frequency = {}

for x in numbers:
    frequency[x] = frequency.get(x, 0) + 1

most_frequent = max(frequency, key=frequency.get)

print("Most frequent number:", most_frequent)
print("Frequency:", frequency[most_frequent])
```

### 4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for each digit} and set B = {digit × 9 for each digit}. Print both sets.
### vi. Find the union of A and B — all unique values across both sets.
### vii. Find the intersection of A and B — values common to both.
### viii. Find A − B and B − A separately (not combined) using difference(), and explain in one line how this differs from symmetric_difference().
### ix. Find the symmetric difference of A and B.
### x. Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset().
### xi. Ask the user to input a value X. Remove X from set A if present using discard() (not remove()) — explain in one line why discard() is safer than remove() when you're not sure the value exists.

```python
roll_no = input("Enter your roll number: ")
digits = [int(digit) for digit in roll_no]
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)
union = A.union(B)
print("Union:", union)

intersection = A.intersection(B)
print("Intersection:", intersection)

A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

symmetric = A.symmetric_difference(B)
print("Symmetric Difference:", symmetric)

print("Is A subset of B?", A.issubset(B))
print("Is B superset of A?", B.issuperset(A))

X = int(input("Enter value to remove from A: "))

A.discard(X)

print("A after discard:", A)
```

```python


my_dict = {
    "name": "Your Name",
    "roll_no": "Your Roll Number",
    "branch": "CSE",
    "age": 20,
    "city": "Your City"
}

print("Original dictionary:")
print(my_dict)


my_dict["location"] = my_dict.pop("city")

print("\nAfter renaming city to location:")
print(my_dict)

my_dict["cgpa"] = 8.5

print("\nAfter adding CGPA:")
print(my_dict)

my_dict["age"] = my_dict["age"] + 1

print("\nAfter increasing age:")
print(my_dict)

dict_pop = my_dict.copy()

removed_branch = dict_pop.pop("branch")

print("\nUsing pop():")
print(dict_pop)
print("Removed value:", removed_branch)

dict_del = my_dict.copy()

del dict_del["branch"]

print("\nUsing del:")
print(dict_del)

print("\nKey-value pairs:")

for key, value in my_dict.items():
    print(key, "→", value)

if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key does not exist.")

friend_dict = {
    "name": "Rahul",
    "roll_no": "12345",
    "branch": "CSE",
    "age": 21,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print("\nMerged dictionary:")
print(merged_dict)
string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nOnly string values:")
print(string_values)
```
