### Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly spaced 25 numbers from 10 to 100 using linspace(). Can we do the same with T attribute?


```python
Kriti=np.linspace(10,100,25)
print(Kriti)
```

```python
print("dimension of array:",Kriti.ndim)
print("Shape of array:",Kriti.shape)
print("Total elements of array:",Kriti.size)
print("datatype of each element in array:",Kriti.dtype)
print("total byte of array:",Kriti.nbytes)
```

```python
T=Kriti.reshape(5,5,order='F')
print(T)
```

```python
print("dimension of array:",T.ndim)
print("Shape of array:",T.shape)
print("Total elements of array:",T.size)
print("datatype of each element in array:",T.dtype)
print("total byte of array:",T.nbytes)
```
